# Random Survival Forest

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline

from sklearn import set_config
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder

from sksurv.ensemble import RandomSurvivalForest

set_config(display="text")  # displays text representation of estimators
```

```python
# Load the dataset.
df_train = pd.read_csv("../data/preprocessed/with_unknown.csv")
df_test = pd.read_csv("../data/test.csv")
```

```python
# Fill the missing values of the test set with 'unknown'
df_test.fillna("Unknown",  inplace=True)
df_test.isnull().sum()
```

```python
# One-hot encode categorical columns in 0/1 format.
df_train = pd.get_dummies(df_train, dtype=int)
df_test = pd.get_dummies(df_test, dtype=int)
```

```python
X_train, X_test = df_train.drop(['ID', 'efs', 'efs_time'], axis=1), df_test.copy()
y = np.array([tuple(row) for row in df_train[['efs', 'efs_time']].values], dtype=[('cens', '?'), ('time', '<f8')])
```

```python
random_state = 20
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_train, y, test_size=0.25, random_state=random_state)
```

```python
rsf = RandomSurvivalForest(n_estimators=100,
                           min_samples_split=10,
                           min_samples_leaf=15,
                           n_jobs=-1,
                           random_state=random_state)
rsf.fit(X_train, y_train)
```

```python
# Check how well the model performs on the test data.
rsf.score(X_test, y_test)
```

## Predicting 

Estimation of the survival and cumulative hazard function using Kaplan-Meier and Nelson-Aalen estimator, respectively. Moreover, risk score can be computed representing the number of events for one particular terminal node. Ensemble prediciton is simply the average across all trees in the forest.
